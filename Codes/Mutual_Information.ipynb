```python
import numpy as np
import pandas as pd
from algorithm import pairwise_mutual_information

import warnings
warnings.filterwarnings("ignore")

# All warnings are now hidden for the duration of this script
```

```python
rng = np.random.default_rng(42)
x = rng.normal(size=500)

df = pd.DataFrame(
    {
        "x": x,
        "y": x + rng.normal(scale=0.2, size=500),
        "group": np.where(x > 0, "positive", "negative"),
        "category_code": rng.integers(0, 3, size=500),
    }
)

mi = pairwise_mutual_information(
    df,
    discrete_columns=["category_code"],
    n_neighbors=3,
    random_state=42,
)

display(mi.round(3))
display(mi.attrs)
```

```python
data = pd.read_excel('../Data/Test Data.xlsx', sheet_name = 'data')
analysis_data = data.drop(columns=["SITE_ID"], errors="ignore")

mi_df = pairwise_mutual_information(
    analysis_data,
    discrete_columns=["PFT", "KoeppenGeiger"],
    missing="drop",
    n_neighbors=3,
    random_state=42,
)

display(mi_df)

# View the number of observations used for each estimate.
sample_sizes = pd.DataFrame(
    mi_df.attrs["pair_sample_sizes"],
    index=mi_df.index,
    columns=mi_df.columns,
)
display(sample_sizes)

display("Singleton categories:", mi_df.attrs["singleton_categories"])
display("Skipped pairs:", mi_df.attrs["skipped_pairs"])
```
